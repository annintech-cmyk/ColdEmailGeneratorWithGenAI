```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name="test")
print(client.list_collections())
```

```python
collection.add(
    documents=["This document is about New York.", "This document is about France."],
    ids=["id1", "id2"]
)
```

```python
all_documents = collection.get()
print(all_documents)
```

```python
documents = collection.get(ids=["id1"])
print(documents)
```

```python
results = collection.query(
    query_texts=["Query is about apple?"],
    n_results=2
)
print(results)
```

sematic search

```python
results1 = collection.query(
    query_texts=["Query is about chromosome?"],
    n_results=2
)
print(results1)
```

Delete All data

```python
collection.delete(ids=["id1"])
collection.delete(ids=["id2"])
collection.get()
```

Create collection with metadata

```python
collection.add(documents=["This document is about New York.", "This document is about France."], ids=["id1", "id2"], metadatas=[{"url": "https://en.wikipedia.org/wiki/New_York_City"}, {"url": "https://en.wikipedia.org/wiki/France"}] )
```

This result will give metadata whuch help us to findout which source refering to 

```python
results = collection.query(
    query_texts=["Query is about apple?"],
    n_results=2
)
print(results)
```
